Loop through: This is how we actually read files in real life

```python
with open("demo.txt", "r") as f:
    for line in f:
        print(line)
```

```python
with open("demo.txt", "r") as f:
    print(f.tell())  # 0
    f.read(6)
    print(f.tell())  # 6
```

seek moves pointer back

```python
with open("demo.txt", "r") as f:
    f.read(6)
    print(f.tell())
    
    f.seek(0)
    print(f.read(5))
```

```python
with open("demo.txt", "r") as f:
    # f.read(6)
    print(f.tell())
    
    f.seek(2)
    print(f.read())
```

Truncate: Cuts file content after 10 characters

```python
with open("demo.txt", "r+") as f:
    f.truncate(10)
```

isatty() checks whether input/output is coming from a terminal (keyboard/screen) or from a file

```python
with open("demo.txt", "r") as f:
    print(f.isatty())
```

```python
def count_M_lines():
    count = 0
    with open("introduction.txt", "r") as file:
        for line in file:
            if line.startswith('M'):   # check if line begins with M
                count += 1
    print("Number of lines starting with M:", count)

# Call the function
count_M_lines()


```
