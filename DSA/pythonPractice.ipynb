Given an integer n. You need to recreate the pattern given below for any value of N. Let's say for N = 5, the pattern should look like as below:


```
**********
****  ****
***    ***
**      **
*        *
*        *
**      **
***    ***
****  ****
**********
```


Print the pattern in the function given to you.

````python
# my sol
class Solution:
    def pattern19(self, n):
        for i in range(n):
            for j in range(0, n-i):
                print("*", end=(""))
            for j in range(2*i):
                print(" ", end=(""))
            for j in range(n-i):
                print("*", end=(""))
            print()
        for i in range(n):
            for j in range(i+1):
                print("*", end=(""))
            for j in range((2*n-2)-2*i):
                print(" ", end=(""))
            for j in range(i+1):
                print("*", end=(""))
            print()
````

````python
class Solution:
    def pattern19(self, n):
        space = 0

        for i in range(n):
            print("*" * (n-i), end=(""))
            print(" " * space, end=(""))
            print("*" * (n-i))
            space += 2

        sp = 2*n-2
        for i in range(1,n+1):
            print("*" * i, end=(""))
            print(" " * sp, end=(""))
            print("*" * i)
            sp -= 2
````
