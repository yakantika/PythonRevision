```python
from abc import ABC, abstractmethod

# abstract base class
class Vehicle(ABC):
    def drive(self):
        print("the vehicle is used for driving")
    
    @abstractmethod
    def start_engin(self):
        pass

class Car(Vehicle):
        def start_engin(self):
            print("the car engine is started")
        
def operate_engin(vehicle):
    vehicle.start_engin()

# car = Car()
car = Vehicle()
operate_engin(car)
```
