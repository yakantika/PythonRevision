```python
import numpy as np

array = np.array([1, 2, 3, 4, 5,6])
print(array)
```

```python
# converted to 2D array
arr = array.reshape(2,3) 
```

```python
# used ones
np.ones((2,3))
```

```python
print("Array: ",arr) #to print arrays
print("Shape: ", arr.shape) #to print shape of array
print("number of dimensions: ", arr.ndim) #to print number of dimensions
print("Number of elements: ", arr.size) #to print number of elements
print("Data type:", arr.dtype) #to print data type of array
print("Size of each element in bytes: ", arr.itemsize) #to print size of each element in bytes
```

```python
# numpy vactorised operation
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

# elementwise addition
print("Sum: ", arr1+arr2)

# elemenwise subtraction
print("subs: ", arr2 - arr1)

# similarly we can do / and *
```

```python
# universal functions
arr = np.array([2,3,4])
print("Square root: ", np.sqrt(arr))
print("Exponential: ", np.exp(arr))
print("Sine: ", np.sin(arr))
print("Natural log:", np.log(arr))
```

```python
arr = np.array([[1,2,3,6],[4,5,6,4],[7,8,9,7]])
print(arr)
```

```python
# to fetch perticular element
print(arr[0,1]) # element at 0th row and 1st column
```

```python
# if we want to fetch 2nd row and 3rd column
print(arr[1:,2:]) # 1: means from 1st row to end and 2: means from 2nd column to end
```

```python
# modify aray element
arr[0,0] = 100
print(arr)
```

```python
# statistical concept - normalization
# to have a mean of 0 and standard deviation of 1
data = np.array([1,2,3,4,5])

# calculating mean and standard daviation
mean = np.mean(data)
std =  np.std(data)

print("Mean: ", mean)
print("standard deviation: ", std)

# normalize the data
normalized_data = (data - mean) / std
print("Normalized data: ", normalized_data)

# variance
variance = np.var(data)
print("Variance: ", variance)

# median
median = np.median(data)
print("median: ", median)
```
