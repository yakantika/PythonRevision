```python
# pandas series is a 1d array like object that can holde and data type. it is similer to column in a table
import pandas as pd

data = [1,2,3,4,5]
series = pd.Series(data)
print("series: \n",series)
```

```python
data = {"a":1,"b":2,"c":3}
series_dict = pd.Series(data)
print("series: \n",series_dict)

# for dictionary all the keys will be treated as index and values will be treated as data
```

```python
data = [10,20,30]
index = ["a","b","c"]
pd.Series(data , index= index)
```

```python
# dataframe is a 2d table like data structure in pandas. it is similer to excel sheet or sql table, having multilple rows and columns
df = pd.DataFrame(data={"name":["Alice","Bob","Charlie"],"age":[25,30,35],"city":["NY","LA","SF"]})
print("dataframe: \n",df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
data = [{"name":"Alice","age":25,"city":"NY"},{"name":"Bob","age":30,"city":"LA"},{"name":"Charlie","age":35,"city":"SF"}]
df_from_dict = pd.DataFrame(data)
print("dataframe from list of dictionaries: \n",df_from_dict)
```
