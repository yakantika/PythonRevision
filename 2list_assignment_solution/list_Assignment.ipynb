### Assignment 6: List Methods

Create a list of random numbers and sort it in ascending and descending order. Remove the duplicates from the list and print the modified list.


```python
import random

rn = [random.randint(1,20) for _ in range(15)]
print(f"original list: {rn}")

sn = sorted(rn)
print(f"sorted list: {sn}")

snd = sorted(rn, reverse=True)
print(f"sorted in descending order: {snd}")

un = list(set(rn))
print(f"uniue list:  {un}")
```
