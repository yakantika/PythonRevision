**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
c = int(input("Enter a year: "))
if (c%4==0 and c%100 != 0) or (c%400 == 0):
    print(f"{c} is a leap year")
else:
    print(f"{c} is not a leap year")
```
