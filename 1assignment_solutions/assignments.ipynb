### Assignment 8: break Statement

Write a program that asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.


```python
tot = 0
while True:
    num = int(input("Enter num: "))
    if num == 0:
        break
    tot += num
print("Sum:", tot)
```

```python
# for i in range (1,11):
    
#     if i==5:
#         continue
#     print(i)

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```
