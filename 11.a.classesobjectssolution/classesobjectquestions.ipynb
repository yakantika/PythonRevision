```python
# 4
class BankAccount:
    def __init__(self, __account_number, __balance):
        self.__account_number = __account_number
        self.__balance = __balance
        
    def withdaw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"withdrawal successfully, available balance: {self.__balance}")

    def deposite(self, amount):
        self.__balance += amount
        print(f"deposite successfully, available balance: {self.__balance}")

    def currnent_balance(self):
        return self.__balance

user1 = BankAccount("123", 1000)
user1.withdaw(500)
user1.deposite(200)
print(user1.currnent_balance())
```

```python
# 5
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

# If the child class defines its own __init__, Python does NOT automatically call the parent constructor.So the parent attributes (name, age) will not exist unless you call it manually.
class Employee(Person):
        def __init__(self, name, age,employee_id):
            super().__init__(name, age)
            self.employee_id = employee_id

        def __str__(self):
            return f"Employee(Name: {self.name}, Age: {self.age}, Empoyee ID: {self.employee_id})"
emp = Employee("isha",24,"123")
print(emp.employee_id)
emp.__str__
```
